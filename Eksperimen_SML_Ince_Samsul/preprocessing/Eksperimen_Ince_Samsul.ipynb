## Perkenalan Dataset

Dataset yang digunakan adalah **Breast Cancer Wisconsin Diagnostic** dari `sklearn.datasets.load_breast_cancer`. Dataset ini berasal dari UCI Machine Learning Repository dan tersedia langsung melalui scikit-learn sehingga proses eksperimen dapat direproduksi tanpa akses eksternal.

Dataset berisi 569 sampel hasil diagnosis tumor payudara dengan 30 fitur numerik, seperti radius, texture, perimeter, area, smoothness, compactness, concavity, dan symmetry. Label target memiliki dua kelas: `0` untuk malignant dan `1` untuk benign. Tujuan eksperimen adalah membangun model klasifikasi untuk memprediksi target diagnosis berdasarkan fitur numerik tersebut.

Tahap preprocessing pada notebook ini mencakup pemuatan dataset mentah, pemisahan fitur dan label, normalisasi fitur menggunakan `StandardScaler`, serta penyimpanan dataset hasil preprocessing agar dapat digunakan ulang pada tahap modeling dan workflow CI.


```python
from pathlib import Path
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
```

```python
data = load_breast_cancer(as_frame=True)
df = data.frame.copy()
df.head()
```

```python
X = df.drop(columns=['target'])
y = df['target']
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
processed = X_scaled.copy()
processed['target'] = y.values
processed.to_csv('breast_cancer_preprocessing.csv', index=False)
processed.head()
```
